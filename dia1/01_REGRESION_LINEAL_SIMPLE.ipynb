## 1. regresion lineal simple de scipy

## 1.1 ejemplo sencilo con datos simulados

```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import scipy
```

```python
import scipy.stats as stats
```

```python
x = np.array([1,2,3,4,5,6,7])
y = np.array([2,2,3,8,8,9,9])
```

```python
print(x,y)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y)
plt.show()
```

## realizamos la regeresion lineal


```python
result = scipy.stats.linregress(x,y)
result
```

# mostramos resultados d ela regresion lineal

```python
print (f'Pendiente (slope) : {result.slope}')
print (f'Intercepto (intercept) : {result.intercept}')
print (f'Coeficiente de correlacion (rvalue) : {result.rvalue}')
print (f'Valor p (pvalue) : {result.pvalue}')
print (f'Error estandar de la pendiente (stderr) : {result.stderr}')
```

```python
y_pred = result.intercept + result.slope*x
y_pred
```

## graficamos la regresion lineal

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y,color = 'blue',label = 'Datos Observados')
plt.plot(x,result.intercept + result.slope*x,color='red',label = 'Regresion Lineal')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## PROBRAMOS NUESTRA PEDICCION

```python
x_new = 5
y_new_pred = result.intercept + result.slope*x_new
print (f'Prediccion para x = {x_new}: y = {y_new_pred}')
```
